```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df["Month"] = df["Date"].dt.strftime("%b")
df["Month"] = pd.Categorical(df["Month"], categories=month_order, ordered=True)

monthly_downtime = df.groupby("Month", observed=True)["Downtime_min"].sum()

monthly_downtime.plot(kind="bar")

plt.xlabel("Month")
plt.ylabel("Total Downtime (min)")
plt.title("Monthly Total Downtime")
plt.xticks(rotation=45)
plt.show()
```

```python
yield_data = df.pivot_table(values="Product_Yield_pct",index="Month",columns="Shift",aggfunc="mean")

yield_data = yield_data.reindex(month_order)

plt.figure(figsize=(8, 6))

sns.heatmap(yield_data,annot=True,fmt=".2f",cmap="YlGnBu")

plt.title("Month-wise and Shift-wise Average Product Yield")
plt.xlabel("Shift")
plt.ylabel("Month")
plt.show()
```

```python
month_order = ["January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"]

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Month"] = df["Date"].dt.month_name()

month_shift_purity = df.pivot_table(
    values="Product_Purity_pct",
    index="Month",
    columns="Shift",
    aggfunc="mean")

month_shift_purity = month_shift_purity.reindex(month_order)

plt.figure(figsize=(10, 7))

sns.heatmap(month_shift_purity,
    annot=True,
    fmt=".2f",
    cmap="Greens",
    linewidths=0.5,
    cbar_kws={"label": "Average Product Purity (%)"})

plt.title("Month-wise and Shift-wise Average Product Purity")
plt.xlabel("Shift")
plt.ylabel("Month")
plt.xticks(rotation=0)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
